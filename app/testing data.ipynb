```python
import os

import numpy as np
import pandas as pd
from features import *
import matplotlib.pyplot as plt
```

```python
def complete_dict(list_dataframe, clean):
    if clean:
        path = "../Data/train/with_label/clean"
    else:
        path = "../Data/train/with_label/dirty"
    for e in os.scandir(path):
        if e.is_file():
            dict = {}
            path = e.path
            dict['name'] = e.name
            features_img = extract_features(path)
            for key in features_img.keys():
                dict[key] = features_img[key]
            if clean:
                dict['class'] = 'clean'
            else:
                dict['class'] = 'dirty'
            list_dataframe.append(dict)
```

```python
def get_dataframe() -> list:
    list_dataframe = []

    complete_dict(list_dataframe, 1)
    complete_dict(list_dataframe, 0)

    return list_dataframe
```

# Tests sur des filtres

```python
filter = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])
```

```python
from PIL import Image
img = Image.open("../Data/train/with_label/dirty/train_no_label_img_521.jpeg")
array_img = np.array(img)
array_img.shape
```

```python
def apply_filter_rgb(array_image, filter):
    height_image, width_image, RGB = array_image.shape
    width_image -=2
    height_image -= 2

    windows = np.lib.stride_tricks.sliding_window_view(array_image, (3, 3), axis=(0, 1))
    print(windows.shape)
    result_r = np.sum(windows[:, :, :, :,0] * filter[np.newaxis, np.newaxis, :, :],axis=(-1, -2))
    result_g = np.sum(windows[:, :, :, :,1] * filter[np.newaxis, np.newaxis, :, :],axis=(-1, -2))
    result_b = np.sum(windows[:, :, :, :,2] * filter[np.newaxis, np.newaxis, :, :],axis=(-1, -2))

    result1 = np.sum(np.array([result_r, result_g, result_b]), axis=0)

    result2 = np.sum(windows * filter[np.newaxis, np.newaxis, np.newaxis, :, :],axis=(-1, -2))
    return result1, result2
```

```python
result1, result2 = apply_filter_rgb(array_img, filter)
result1.shape
```

```python
final_image_array = np.clip(result2, 0, 255).astype(np.uint8)
img_after_filter = Image.fromarray(final_image_array, mode="RGB")

img_after_filter.show()
```

```python
def extract_features_images(img) -> dict:
    """
    Taille de l'image : en ko
    Dimension de l'image: width x height en pixels
    Moyenne RGB : 3 canaux
    Luminosité RGB: basé sur le calcul REC601 (voir discord)
    Contraste de couleur: max(pixel) - min(pixel)
    Saturation : saturation moyenne HSV (0–255)
    Histogramme niveaux de gris : 256 valeurs (JSON)
    Histogramme RGB       : r/g/b × 256 valeurs (JSON)
    Histogramme luminance : 256 valeurs (JSON)
    Densité de contours   : % de pixels avec un bord détecté (0–1)

    :param file_path: uploads/...
    :return: dictionary with features of a specific image
    """
    features = dict()


    features["image_width"], features["image_height"] = img.size

    # RGB, Couleur moyenne
    stat = ImageStat.Stat(img)
    features["mean_r"] = round(stat.mean[0],2)
    features["mean_g"] = round(stat.mean[1],2)
    features["mean_b"] = round(stat.mean[2],2)


    # Luminosité, valeur entre 0 et 255
    features["brightness"] = features["mean_r"] * 0.299 + features["mean_g"] * 0.587 + features["mean_b"] * 0.114

    # gray image
    gray = img.convert("L")
    extrema = gray.getextrema()
    # Niveau de contraste
    features["contraste"] = extrema[1] - extrema[0]

    # Saturation moyenne
    img_hsv = img.convert("HSV")
    stat_hsv = ImageStat.Stat(img_hsv)
    features["saturation"] = round(stat_hsv.mean[1], 2)

    # Histogrammes RGB
    r, g, b = img.split()
    r_histo = r.histogram()
    g_histo = g.histogram()
    b_histo = b.histogram()
    features["hist_rgb"] = json.dumps({
        "red":   r_histo,
        "green": g_histo,
        "blue":  b_histo,
    })

    # Histogramme luminance
    features["hist_luminance"] = json.dumps(gray.histogram())


    # Contours de l'image
    # % en fonction des pixels actifs (bords détectés)
    # Valeur élevée = image complexe/chargée (poubelle pleine)
    edge_img = gray.filter(ImageFilter.FIND_EDGES)
    stat_edge = ImageStat.Stat(edge_img)
    features["edge_density"] = round(stat_edge.mean[0] / 255.0, 4)

    return features
```

```python
my_dict = extract_features_images(img_after_filter)

if isinstance(my_dict.get('hist_rgb'), str):
    my_dict['hist_rgb'] = json.loads(my_dict['hist_rgb'])
if isinstance(my_dict.get('hist_luminance'), str):
    my_dict['hist_luminance'] = json.loads(my_dict['hist_luminance'])

# 2. Convert to DataFrame by wrapping the dict in a list []
df_after_filter = pd.DataFrame([my_dict])
```
